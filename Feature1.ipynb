```python
from pyspark.sql import functions as F

# Step 1: ดึง schema จาก 1 ไฟล์ก่อน (แก้ปัญหา UNABLE_TO_INFER_SCHEMA)
schema = spark.read.parquet(
    "/Volumes/workspace/default/mule_raw/transactions/batch-3/part_0297.parquet"
).schema

# Step 2: โหลดทุก batch พร้อมกันโดยใช้ schema ที่ได้
df = spark.read \
    .schema(schema) \
    .option("recursiveFileLookup", "true") \
    .parquet("/Volumes/workspace/default/mule_raw/transactions/")

# Step 3: แปลง timestamp และกรอง null
df = df.withColumn("transaction_timestamp", F.to_timestamp("transaction_timestamp")) \
       .filter(F.col("account_id").isNotNull()) \
       .filter(F.col("transaction_timestamp").isNotNull()) \
       .filter(F.col("amount").isNotNull()) \
       .filter(F.col("txn_type").isNotNull())



# Step 5: ตรวจสอบ
print(f"Total rows : {df.count():,}")
print(f"Columns    : {df.columns}")
display(df.limit(5))
```

```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

# ---- 1. เตรียมข้อมูล ----
credits = df.filter((F.col("txn_type") == "C") & (F.col("amount") > 0)) \
            .withColumnRenamed("transaction_timestamp", "credit_ts") \
            .withColumnRenamed("amount", "credit_amount") \
            .select("account_id", "credit_ts", "credit_amount")

debits = df.filter((F.col("txn_type") == "D") & (F.col("amount") > 0)) \
           .withColumnRenamed("transaction_timestamp", "debit_ts") \
           .withColumnRenamed("amount", "debit_amount") \
           .select("account_id", "debit_ts", "debit_amount")

# ---- 2. Join credit x debit ในบัญชีเดียวกัน (debit ต้องเกิดหลัง credit และใน 24h) ----
joined = credits.join(
    debits,
    on=[
        credits["account_id"] == debits["account_id"],
        debits["debit_ts"] > credits["credit_ts"],
        debits["debit_ts"] <= credits["credit_ts"] + F.expr("INTERVAL 24 HOURS")
    ],
    how="left"
) \
.select(
    credits["account_id"],
    "credit_ts",
    "credit_amount",
    "debit_ts",
    "debit_amount"
)

# ---- 3. คำนวณ % debit เทียบกับ credit ก้อนนั้นๆ แต่ละช่วงเวลา ----
joined = joined.withColumn(
    "debit_pct",
    F.coalesce(F.col("debit_amount") / F.col("credit_amount") * 100, F.lit(0))
)

# รวม % debit ต่อ (account_id, credit_ts) แยก 3 window
per_credit = joined.groupBy("account_id", "credit_ts", "credit_amount") \
    .agg(
        # ด่าน 1: % ที่โอนออกใน 15 นาทีแรก
        F.sum(
            F.when(
                F.col("debit_ts") <= F.col("credit_ts") + F.expr("INTERVAL 15 MINUTES"),
                F.col("debit_pct")
            ).otherwise(0)
        ).alias("pct_out_15min"),

        # ด่าน 2: % ที่โอนออกในช่วง 16นาที - 1 ชั่วโมง
        F.sum(
            F.when(
                (F.col("debit_ts") > F.col("credit_ts") + F.expr("INTERVAL 15 MINUTES")) &
                (F.col("debit_ts") <= F.col("credit_ts") + F.expr("INTERVAL 1 HOURS")),
                F.col("debit_pct")
            ).otherwise(0)
        ).alias("pct_out_1hr"),

        # ด่าน 3: % ที่โอนออกรวมใน 24 ชั่วโมง
        F.sum(F.col("debit_pct")).alias("pct_out_24hr")
    )

# ---- 4. ตรวจสอบแต่ละด่าน → Flag 0/1 ----
per_credit = per_credit \
    .withColumn("flag_15min", F.when(F.col("pct_out_15min") >= 28, 1).otherwise(0)) \
    .withColumn("flag_1hr",   F.when(F.col("pct_out_1hr")   >= 25, 1).otherwise(0)) \
    .withColumn("flag_24hr",  F.when(F.col("pct_out_24hr")  >= 85, 1).otherwise(0))

# ---- 5. Aggregate → นับความถี่ต่อบัญชี (Frequency Index) ----
final_features = per_credit.groupBy("account_id") \
    .agg(
        F.sum("flag_15min").alias("count_rapid_15min"),   # นับครั้งที่เข้าเกณฑ์ด่าน 1
        F.sum("flag_1hr").alias("count_rapid_1hr"),       # นับครั้งที่เข้าเกณฑ์ด่าน 2
        F.sum("flag_24hr").alias("count_drain_24hr"),     # นับครั้งที่เข้าเกณฑ์ด่าน 3
        F.count("credit_ts").alias("total_credit_events") # รับเงินทั้งหมดกี่รอบ
    ) \
    .withColumn(
        "frequency_index",  # คะแนนรวมทุกด่าน
        F.col("count_rapid_15min") + F.col("count_rapid_1hr") + F.col("count_drain_24hr")
    )

display(final_features.orderBy(F.col("frequency_index").desc()))
```
